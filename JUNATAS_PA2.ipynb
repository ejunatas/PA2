```python
import numpy as np
```

### A:

```python
np.random.seed(2112) #Set seed for 5x5 array
X = np.random.randint(10, 101, size=(5,5))

mean_X = np.mean(X) #Solving for mean
std_X = np.std(X) #Solving for standard deviation
X_normalized = (X - mean_X) / std_X # For normalization of the array

# Printing array, mean, standard deviation, and the normalized array
print("Array:")
print(X)
print("Normalized Array")
print(X_normalized)
print("Mean of X_normalize (must be 0):", X_normalized.mean())
print("Standard Deviation of X_normalize (must be 1):", X_normalized.std())

#Saving the array into a file
np.save('X_normalized.npy', X_normalized)
```

### B:

```python
integers = np.arange(1, 101) # Create first 100 positive integers
C = (integers ** 3).reshape(10, 10) # Cube every element and reshape to 10x10
div_by_4 = C[C % 4 == 0] # Select elements divisible by 4

#Printing array, number of elements, first element, and last element
print("Shape of array C:", C.shape)
print("Array divide_by_4:")
print(divide_by_4)
print("Number of selected elements:", divide_by_4.size)
print("First element:", divide_by_4[0])
print("Last element:", divide_by_4[-1])

# Save the selected array
np.save('div_by_4.npy', div_by_4) 
```

### C:

```python
squares = np.arange(1, 37) ** 2  # Creating a 6x6 array with the square of the first 36 positive integers.
S = squares.reshape(6, 6)   

S_mean = np.mean(S) # Mean of array

above_mean = S[S > S_mean] #Select integers that are only above the mean

# Printing array, mean, integers above the mean, number of selected integers, first element, and last element.
print("Array S:")
print(S)
print("\nMean of S (S_mean):", S_mean)
print("\nArray above_mean:")
print(above_mean)
print("\nNumber of selected elements:", above_mean.size)
print("First element:", above_mean[0])
print("Last element:", above_mean[-1])

# Saving array into a file
np.save('above_mean.npy', above_mean)
```
